## Embeddings

1. **Word Embeddings:** esses são talvez o tipo mais comum de embeddings, amplamente usados no processamento de linguagem natural (PNL). Modelos como Word2Vec, GloVe e FastText criam representações vetoriais densas de palavras, onde palavras com significados semelhantes têm representações semelhantes. Esses vetores capturam relações semânticas e nuances de contexto.

## Seq-2-Seq

Seq2Seq é um modelo ou uma arquitetura de rede neural projetada para transformar sequências de entrada em sequências de saída. Ele é comumente usado para tarefas como tradução automática, onde a sequência de entrada (texto em um idioma) precisa ser convertida para uma sequência de saída (texto em outro idioma).

* **Propósito:** Converter uma sequência de entrada em uma sequência de saída onde as duas sequências podem ser de comprimentos diferentes. É particularmente útil em tarefas onde o contexto completo da entrada é necessário para gerar a saída.
* **Componentes:** Um modelo seq2seq típico consiste em duas partes principais - um encoder e um decoder. O encoder lê a sequência de entrada e condensa a informação em um vetor de contexto. O decoder então usa esse vetor para gerar a sequência de saída passo a passo.
* **Uso comum:** Além de tradução de linguagem, seq2seq é usado em tarefas como resumo de texto, onde uma longa entrada é condensada em um texto mais curto, e em chatbots, onde as entradas do usuário são convertidas em respostas conversacionais.

### Tensorflow

```
import numpy as np
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Dense
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Dados de exemplo
english = ['cat', 'dog', 'yesterday']
spanish = ['gato', 'perro', 'ayer']

# Vocabulário manual para simplificação
word_to_idx_eng = {'cat': 1, 'dog': 2, 'yesterday': 3, '0': 0}
word_to_idx_spa = {'gato': 1, 'perro': 2, 'ayer': 3, '0': 0}

# Converter palavras para índices
english_seq = [[word_to_idx_eng[word]] for word in english]
spanish_seq = [[word_to_idx_spa[word]] for word in spanish]

# Parâmetros
num_encoder_tokens = len(word_to_idx_eng) + 1
num_decoder_tokens = len(word_to_idx_spa) + 1

# Converter índices para sequências de one-hot encoding
encoder_input_data = to_categorical(pad_sequences(english_seq, maxlen=1, padding='post'), num_classes=num_encoder_tokens)
decoder_input_data = to_categorical(pad_sequences(spanish_seq, maxlen=1, padding='post'), num_classes=num_decoder_tokens)

# Ajuste: O target deve ser deslocado por uma posição no tempo para o treinamento do decoder
decoder_target_data = np.roll(decoder_input_data, -1, axis=1)

# Corrigir as dimensões adicionando a dimensão de timestep se necessário
encoder_input_data = np.expand_dims(encoder_input_data, 1)  # Formato esperado: (batch_size, seq_length, num_tokens)
decoder_input_data = np.expand_dims(decoder_input_data, 1)
decoder_target_data = np.expand_dims(decoder_target_data, 1)

# Configurações da rede
latent_dim = 256
encoder_inputs = Input(shape=(None, num_encoder_tokens), name='encoder_inputs')
encoder = LSTM(latent_dim, return_state=True, name='encoder_lstm')
encoder_outputs, state_h, state_c = encoder(encoder_inputs)
encoder_states = [state_h, state_c]

decoder_inputs = Input(shape=(None, num_decoder_tokens), name='decoder_inputs')
decoder_lstm = LSTM(latent_dim, return_sequences=True, return_state=True, name='decoder_lstm')
decoder_outputs, _, _ = decoder_lstm(decoder_inputs, initial_state=encoder_states)
decoder_dense = Dense(num_decoder_tokens, activation='softmax', name='decoder_dense')
decoder_outputs = decoder_dense(decoder_outputs)

# Modelo final
model = Model([encoder_inputs, decoder_inputs], decoder_outputs)

# Compilação do modelo
model.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['accuracy'])

# Definição do modelo do encoder para uso na inferência
encoder_model = Model(encoder_inputs, encoder_states)

# Definição do modelo do decoder para uso na inferência
decoder_state_input_h = Input(shape=(latent_dim,), name='decoder_state_input_h')
decoder_state_input_c = Input(shape=(latent_dim,), name='decoder_state_input_c')
decoder_states_inputs = [decoder_state_input_h, decoder_state_input_c]
decoder_outputs, state_h_dec, state_c_dec = decoder_lstm(
    decoder_inputs, initial_state=decoder_states_inputs)
decoder_states = [state_h_dec, state_c_dec]
decoder_outputs = decoder_dense(decoder_outputs)
decoder_model = Model(
    [decoder_inputs] + decoder_states_inputs,
    [decoder_outputs] + decoder_states)

# Treinamento do modelo
model.fit([encoder_input_data, decoder_input_data], decoder_target_data,
          batch_size=1,
          epochs=100,
          validation_split=0.2)
```

```
def translate(input_text, encoder_model, decoder_model, input_token_index, target_token_index, max_decoder_seq_length):
    # Preparação dos índices de entrada
    input_seq = [input_token_index.get(word, 0) for word in input_text.split()]
    input_seq = pad_sequences([input_seq], maxlen=max_decoder_seq_length, padding='post')
    input_seq = to_categorical(input_seq, num_classes=len(input_token_index) + 1)

    # Obtenção dos estados iniciais do encoder
    states_value = encoder_model.predict(input_seq)

    # Inicialização da sequência de saída com o token inicial
    target_seq = np.zeros((1, 1, len(target_token_index) + 1))
    target_seq[0, 0, target_token_index['0']] = 1  # Start token

    decoded_sentence = ''
    stop_condition = False
    while not stop_condition:
        output_tokens, h, c = decoder_model.predict([target_seq] + states_value)

        # Seleção do token com maior probabilidade
        sampled_token_index = np.argmax(output_tokens[0, -1, :])
        sampled_word = [word for word, index in target_token_index.items() if index == sampled_token_index][0]
        if sampled_word == '0':  # Supondo '0' como token de parada
            stop_condition = True
        else:
            decoded_sentence += ' ' + sampled_word

        # Preparação do próximo token de entrada
        target_seq = np.zeros((1, 1, len(target_token_index) + 1))
        target_seq[0, 0, sampled_token_index] = 1

        # Atualização dos estados
        states_value = [h, c]

    return decoded_sentence.strip()
```

```
test_english_texts = ['yesterday']
for text in test_english_texts:
    translated_text = translate(text, encoder_model, decoder_model, word_to_idx_eng, word_to_idx_spa, 10)
    print(f'{text} -> {translated_text}')
```

### Pytorch

```
import torch
from torch.nn.utils.rnn import pad_sequence
from torch import nn, optim
import torch.nn.functional as F
import random

# Vocabulário manual para simplificação
word_to_idx_eng = {'one': 1, 'two': 2, 'three': 3, '0': 0}
word_to_idx_spa = {'uno': 1, 'dos': 2, 'tres': 3, '0': 0}

# Dados de exemplo
english = ['one', 'two', 'three']
spanish = ['uno', 'dos', 'tres']

# Converter palavras para índices
english_seq = [[word_to_idx_eng[word]] for word in english]
spanish_seq = [[word_to_idx_spa[word]] for word in spanish]

# Converter para tensores e pad
english_seq = pad_sequence([torch.tensor(seq) for seq in english_seq], padding_value=0)
spanish_seq = pad_sequence([torch.tensor(seq) for seq in spanish_seq], padding_value=0)

# Definir device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```
class Encoder(nn.Module):
    def __init__(self, input_dim, emb_dim, hidden_dim, n_layers):
        super(Encoder, self).__init__()
        self.hidden_dim = hidden_dim
        self.n_layers = n_layers
        self.embedding = nn.Embedding(input_dim, emb_dim)
        self.rnn = nn.LSTM(emb_dim, hidden_dim, n_layers, batch_first=True)

    def forward(self, src):
        embedded = self.embedding(src)
        outputs, (hidden, cell) = self.rnn(embedded)
        return hidden, cell

class Decoder(nn.Module):
    def __init__(self, output_dim, emb_dim, hidden_dim, n_layers):
        super(Decoder, self).__init__()
        self.output_dim = output_dim
        self.embedding = nn.Embedding(output_dim, emb_dim)
        self.rnn = nn.LSTM(emb_dim, hidden_dim, n_layers, batch_first=True)
        self.fc_out = nn.Linear(hidden_dim, output_dim)

    def forward(self, input, hidden, cell):
        input = input.unsqueeze(1)
        embedded = self.embedding(input)
        output, (hidden, cell) = self.rnn(embedded, (hidden, cell))
        prediction = self.fc_out(output.squeeze(1))
        return prediction, hidden, cell

class Seq2Seq(nn.Module):
    def __init__(self, encoder, decoder, device):
        super(Seq2Seq, self).__init__()
        self.encoder = encoder
        self.decoder = decoder
        self.device = device

    def forward(self, src, trg, teacher_forcing_ratio = 0.5):
        batch_size = trg.shape[0]
        trg_len = trg.shape[1]
        trg_vocab_size = self.decoder.output_dim

        outputs = torch.zeros(batch_size, trg_len, trg_vocab_size).to(self.device)

        hidden, cell = self.encoder(src)

        input = trg[:, 0]

        for t in range(1, trg_len):
            output, hidden, cell = self.decoder(input, hidden, cell)
            outputs[:, t] = output
            teacher_force = random.random() < teacher_forcing_ratio
            top1 = output.argmax(1)
            input = trg[:, t] if teacher_force else top1

        return outputs
```

```
# Definições
INPUT_DIM = len(word_to_idx_eng) + 1
OUTPUT_DIM = len(word_to_idx_spa) + 1
ENC_EMB_DIM = DEC_EMB_DIM = 32
HIDDEN_DIM = 128
N_LAYERS = 1

encoder = Encoder(INPUT_DIM, ENC_EMB_DIM, HIDDEN_DIM, N_LAYERS).to(device)
decoder = Decoder(OUTPUT_DIM, DEC_EMB_DIM, HIDDEN_DIM, N_LAYERS).to(device)
seq2seq = Seq2Seq(encoder, decoder, device).to(device)

optimizer = optim.Adam(seq2seq.parameters())
criterion = nn.CrossEntropyLoss(ignore_index=0)

# Loop de treinamento
for epoch in range(100):
    optimizer.zero_grad()
    output = seq2seq(english_seq.to(device), spanish_seq.to(device))
    output_dim = output.shape[-1]
    output = output[:, 1:].reshape(-1, output_dim)
    spanish_seq_flat = spanish_seq[:, 1:].reshape(-1)
    loss = criterion(output, spanish_seq_flat.to(device))
    loss.backward()
    optimizer.step()
    print(f'Epoch {epoch}, Loss: {loss.item()}')
```

```
idx_to_word_eng = {idx: word for word, idx in word_to_idx_eng.items()}
idx_to_word_spa = {idx: word for word, idx in word_to_idx_spa.items()}
```

```
def translate_sentence(model, sentence, word_to_idx, idx_to_word, device):
    model.eval()  # Evaluation mode
    with torch.no_grad():
        # Tokenizing the sentence, mapping to indices
        input_indices = [word_to_idx.get(word, 0) for word in sentence.lower().split()]
        input_tensor = torch.tensor([input_indices], dtype=torch.long).to(device)

        # Forward pass through the model
        output = model(input_tensor, input_tensor)
        output_indices = output.argmax(2).squeeze(0).tolist()  # Converting output probabilities to indices

        # Convert indices to words
        translated_sentence = ' '.join(idx_to_word.get(idx, '<UNK>') for idx in output_indices if idx != 0)
        return translated_sentence

# Example usage
test_sentence = "one two three"
translated_sentence = translate_sentence(seq2seq, test_sentence, word_to_idx_eng, idx_to_word_spa, device)
print("Translated sentence:", translated_sentence)
```

## Transformers

Transformers são uma classe de modelos de deep learning que têm sido amplamente usados para uma variedade de tarefas de processamento de linguagem natural (NLP). Um exemplo bem conhecido de transformer é o BERT (Bidirectional Encoder Representations from Transformers), desenvolvido pelo Google. Ele é projetado para pré-treinar representações de linguagem profunda analisando o texto de forma bidirecional. Isso permite que o BERT capture o contexto de uma palavra de todas as direções (tanto da esquerda para a direita quanto da direita para a esquerda), o que é uma melhoria significativa em relação aos modelos anteriores que só analisavam o texto em uma direção.

```
from transformers import M2M100ForConditionalGeneration, M2M100Tokenizer

# Carregar o tokenizer e o modelo
model_name = 'facebook/m2m100_418M'
tokenizer = M2M100Tokenizer.from_pretrained(model_name)
model = M2M100ForConditionalGeneration.from_pretrained(model_name)
```

```
# Definir o idioma de origem e de destino
tokenizer.src_lang = "en"
tokenizer.tgt_lang = "fr"
```

```
# Preparar a entrada para o modelo
sentence = "What is the capital of France?"
input_ids = tokenizer(sentence, return_tensors="pt").input_ids

# Gerar a saída pelo modelo
outputs = model.generate(input_ids)
translated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

# Exibir a tradução
print(translated_text)
```
